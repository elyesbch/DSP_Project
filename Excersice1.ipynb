```python
from DataProcessor import DataProcessor
import matplotlib.pyplot as plt
import numpy as np

DEFAULT_DURATION = 4.0
dp = DataProcessor("rawdata/X22/")
files = ['rawdata/X22/Laufdata2.pickle', 'rawdata/X22/schnell_Laufen10.pickle', 'rawdata/X22/rennen1.pickle'];
titles = ['normales Gehen', 'schnelles Gehen ', 'Rennen'];
fs = dp.fs
fig, axes = plt.subplots(6, 1, figsize=(10, 8), sharex=True)

#Hilfsfunktion: eine Messung laden und Zeit + alle drei Achsen zurückgeben
def load_time_and_acc_xyz():
     #Zeitvektor t (in Sekunden) aus dem Acceleration-DataFrame holen
    t = dp.dfAcc["t"].values
    
    num_samples= min(int(DEFAULT_DURATION*fs), len(t))
    
    # 5.6) x-, y- und z-Komponente der Beschleunigung holen
    x = dp.dfAcc["x"].values
    y = dp.dfAcc["y"].values
    z = dp.dfAcc["z"].values
    
    t = t[:num_samples]
    
    x = x[:num_samples]
    y = y[:num_samples]
    z = z[:num_samples]
    
    return t, x, y, z

def load_time_and_gyr_xyz():
     
    t = dp.dfGyr["t"].values
     
    num_samples= min(int(DEFAULT_DURATION*fs), len(t))
    
    #x-, y- und z-Komponente der Beschleunigung holen
    x = dp.dfGyr["x"].values
    y = dp.dfGyr["y"].values
    z = dp.dfGyr["z"].values
    
    t = t[:num_samples]
    
    x = x[:num_samples]
    y = y[:num_samples]
    z = z[:num_samples]

    return t, x, y, z


for i in range(3): 
    fileName = files[i]
    dp.loadRawData(fileName)
    for dev in dp.getDevices():
        dp.loadRawDataDevice(dev)
            
    t1, acc_x, acc_y, acc_z =  load_time_and_acc_xyz()
    t2, gyr_x, gyr_y, gyr_z =  load_time_and_gyr_xyz()
        
    #Drei Unterplots untereinander erstellen
    axes[i].plot(t1, acc_x, label="x")
    axes[i].plot(t1, acc_y, label="y")
    axes[i].plot(t1, acc_z, label="z")
    axes[i].set_ylabel("Acceleration [g]")
    axes[i].set_title(titles[i])
    axes[i].legend()
    axes[i].grid(True)

    axes[i+3].plot(t2, gyr_x, label="x")
    axes[i+3].plot(t2, gyr_y, label="y")
    axes[i+3].plot(t2, gyr_z, label="z")
    axes[i+3].set_ylabel("Gyroscope [deg/s]")
    axes[i+3].set_title(titles[i])
    axes[i+3].legend()
    axes[i+3].grid(True)
    
    axes[5].set_xlabel("Zeit [s]")
    plt.tight_layout()
plt.show()
```
