## Load all files and visualize

```python
from DataProcessor import DataProcessor
from axiamo_lib.Visualizer import Visualizer

dp = DataProcessor("rawdata/X22/Excersice_2/rennen_hakim/Mes_1")

all_data_files = dp.find_all_data_files()
print(all_data_files)
for fileName in all_data_files:
    print("Visualizing file: {fileName}")  
    dp.loadRawData(fileName)
    for dev in dp.getDevices():
        vis = Visualizer()
        print("Processing device: {dev}")
        dp.loadRawDataDevice(dev)
        vis.setDataSet(dp.ds)
        vis.visualizeRawDataSubPlot(dp,fileName)
```

## Load specific file and plot

```python
import matplotlib.pyplot as plt
from DataProcessor import DataProcessor

%matplotlib inline

dp = DataProcessor("rawdata/")
fileName = 'rawdata/X22/rennen1.pickle'
dp.loadRawData(fileName)
for dev in dp.getDevices():
    dp.loadRawDataDevice(dev)

print(dp.ds.dfAcc)           # Pandas DataFrame with all the data from the accelerometer
#print(dp.ds.dfAcc['t'])     # Time vector
#print(dp.ds.dfAcc['x'])     # x axis accelerometer
#print(dp.ds.dfAcc['y'])     # y axis accelerometer
#print(dp.ds.dfAcc['z'])     # z axis accelerometer

plt.figure()
plt.plot(dp.dfAcc['t'], dp.dfAcc['x'], label='X Axis Accelerometer', color='green')
plt.title('X Axis Accelerometer')
plt.xlabel('Time [s]')
plt.ylabel('Acceleration [g]')
plt.legend()
plt.grid(True)
plt.show()
```
